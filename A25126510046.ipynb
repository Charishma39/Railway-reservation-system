```python
import math
import random

print("Math Operations Program")

a = random.randint(1, 100)
b = random.randint(1, 100)

print("Value of a:", a)
print("Value of b:", b)

addition = a + b
print("Addition:", addition)

subtraction = a - b
print("Subtraction:", subtraction)

multiplication = a * b
print("Multiplication:", multiplication)

division = a / b
print("Division:", division)

power = math.pow(a, 2)
print("Square of a:", power)

square_root = math.sqrt(b)
print("Square Root of b:", square_root)

factorial_num = random.randint(1, 10)
factorial_result = math.factorial(factorial_num)

print("Factorial Number:", factorial_num)
print("Factorial:", factorial_result)

angle = random.randint(1, 90)

print("Angle:", angle)

sine_value = math.sin(math.radians(angle))
print("Sine Value:", sine_value)

cos_value = math.cos(math.radians(angle))
print("Cosine Value:", cos_value)

tan_value = math.tan(math.radians(angle))
print("Tangent Value:", tan_value)

log_value = math.log(a)
print("Logarithm of a:", log_value)

ceil_value = math.ceil(division)
print("Ceiling Value:", ceil_value)

floor_value = math.floor(division)
print("Floor Value:", floor_value)

pi_value = math.pi
print("Value of PI:", pi_value)

print("Math Program Completed")
```

```python
import math
import random

print("Python Math Package Program")

x = random.randint(1, 50)
y = random.randint(1, 50)

print("Value of x:", x)
print("Value of y:", y)

sum_result = x + y
print("Addition:", sum_result)

difference = x - y
print("Subtraction:", difference)

product = x * y
print("Multiplication:", product)

quotient = x / y
print("Division:", quotient)

modulus = x % y
print("Modulus:", modulus)

square_x = math.pow(x, 2)
print("Square of x:", square_x)

cube_y = math.pow(y, 3)
print("Cube of y:", cube_y)

sqrt_x = math.sqrt(x)
print("Square Root of x:", sqrt_x)

factorial_value = math.factorial(5)
print("Factorial of 5:", factorial_value)

gcd_value = math.gcd(x, y)
print("GCD:", gcd_value)

angle = 45

sin_value = math.sin(math.radians(angle))
print("Sin 45:", sin_value)

cos_value = math.cos(math.radians(angle))
print("Cos 45:", cos_value)

tan_value = math.tan(math.radians(angle))
print("Tan 45:", tan_value)

log_value = math.log(10)
print("Natural Log:", log_value)

exp_value = math.exp(2)
print("Exponential Value:", exp_value)

ceil_result = math.ceil(quotient)
print("Ceiling:", ceil_result)

floor_result = math.floor(quotient)
print("Floor:", floor_result)

print("Value of PI:", math.pi)
print("Value of Euler:", math.e)

print("Math Package Program Completed")
```

```python
import random
import math
import time
import os

numbers = [random.randint(1, 100) for i in range(10)]

print("Random Numbers:")
print(numbers)

largest = max(numbers)
smallest = min(numbers)

print("Largest:", largest)
print("Smallest:", smallest)

average = sum(numbers) / len(numbers)

print("Average:", average)

square_roots = []

for num in numbers:
    root = math.sqrt(num)
    square_roots.append(root)

print("Square Roots:")
print(square_roots)

current_time = time.ctime()

print("Current Time:")
print(current_time)

folder_name = "sample_folder"

if not os.path.exists(folder_name):
    os.mkdir(folder_name)
    print("Folder Created")
else:
    print("Folder Already Exists")

random_choice = random.choice(numbers)

print("Random Choice:", random_choice)

numbers.sort()

print("Sorted Numbers:")
print(numbers)

for i in range(5):
    print("Python Packages Program", i + 1)

print("Program Finished Successfully")
```
